```python
#4.Write a Python function to find the factorial of 5. Expected Output:
#120

factorialOfNumbers = 1

for i in range(5,1,-1):
    factorialOfNumbers = factorialOfNumbers * i

print(factorialOfNumbers)
```

```python
#5.Write a Python program to check if a number is prime. Expected Output (for input 7):
#7 is a prime number

numTocheck = int(input("Enter number to check prime or not"))
if numTocheck <= 1:
    print(numTocheck, "is not a prime number")
else:
    for i in range(2, (numTocheck//2) +1 ):
        if numTocheck % i == 0 :
            print(numTocheck, "is not a prime number")
            break
    else:
        print(numTocheck, "is a prime number")
```
