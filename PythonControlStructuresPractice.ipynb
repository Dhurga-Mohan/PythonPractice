```python
bmiValue = int(input("Enter the BMI Index:"))
#Calculate the BMI
if(bmiValue < 18.5):
    print("Underweight")
elif(bmiValue >= 18.5 and bmiValue < 24.9):
    print("Healthy weight")
elif(bmiValue >=25 and bmiValue <= 29.9):
    print("Overweight")
elif(bmiValue >= 30):
    print("Very Overweight")
```
